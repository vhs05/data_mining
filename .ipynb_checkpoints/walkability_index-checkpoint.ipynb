```python
start = 25
end = 50
print("Prime numbers between", start, "and", end, "are:")

for num in range(start, end):
    # all prime numbers are greater than 1
    # if number is less than or equal to 1, it is not prime
    if num > 1:
        for i in range(2, num):
            # check for factors
            if (num % i) == 0:
                # not a prime number so break inner loop and
                # look for next number
                break
        else:
            print(num)
```

```python
class Shape:
    def area(self):
        raise NotImplementedError("Area method must be implemented by subclasses")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):

        return 3.14*self.radius**2

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):

        return self.side**2

# Example of polymorphism
shapes = [Circle(5), Square(7), Circle(3)]

for shape in shapes:
    print(shape.area())  # Output: 78.53975, 49, 28.27431
```

```python
# Class Name: Employee, method overriding
class Employee:
    def __init__(self, ID, salary):
        self.ID=ID
        self.salary=salary
    def get_info(self):
        return (f"EmployeeID:{self.ID} Salary{self.salary}")
class SalesEmployee(Employee):
    def __init__(self, ID, salary, sales=0):
        super().__init__(ID, salary)
        self.sales=sales
    def get_info(self):
        return (f"EmployeeID:{self.ID} Salary{self.salary} Sales{self.sales}")
new_employee=Employee(12, 2000)
print(new_employee.get_info())


# Implement the following classes to understand inheritance in Python:

#     Class Name: Employee
#         Attributes:
#             id (int)
#             salary (int)
#         Constructor: __init__(self, id, salary) — Initializes the values to respective variables.
#     Class Name: SalesEmployee (Subclass of Employee)
#         Attributes:
#             Inherited from Employee: id, salary
#             New attribute: sales (int)
        # Constructor: __init__(self, id, salary, sales) — Calls super().__init__(id, salary) to initialize id and salary, and initializes the sales attribute.


```

```python
# Problem: Implement a Python class CustomList that mimics Python's built-in list. It should
# support the following operations:
# • Appending an element
# • Removing an element by value
# • Retrieving an element by index
# • Inserting an element at a specific index
# Ensure that your class handles edge cases, such as removing an element not present in the list or
# accessing an index out of range.

class CustomList:
    def __init__(self):
        self.new_list=[]
    def append(self,num):
        self.new_list.append(num)
    def remove(self,num):
        self.new_list.remove(num)
    def get(self,index):
        self.new_list.get(num)
    def insert(self, index,num):
        self.new_list.insert(index, num)
        
list=CustomList()
print(list.append(12))


# Implement deposit/withdraw/get_balance, then add interest in SavingsAccount.

class BankAccount:
    def __init__(self, account, balance=0):
        self.account=account
        self.balance=balance
    def deposit(self, amount):
        self.balance+=amount
    def withdraw(self, amount):
        if amount <= self.balance:
            return self.balance
        else: 
            print("insufficient funds")
    def get_balance(self):
        return self.balance

class SavingsAccount(BankAccount):
    def __init__(self,account, balance,interest_rate=0.0):
        super().__init__(account, balance)
        self.interest_rate=interest_rate
    def apply_interest(self):
        new_amount=self.balance*self.interest_rate
        return self.balance
        
account = BankAccount("1299", 12)
account.deposit(1500)
account.withdraw(200)
print("Balance:", account.get_balance())
savings = SavingsAccount("1228", 1000, 0.06)
savings.apply_interest()
print("Savings Balance after interest:", savings.get_balance())
    
```

```python
grades = {
"Alice": [85, 90, 88],
"Bob": [78, 82, 80],
"Charlie": [92, 95, 93]
}
# Write a function average_grades(grades) that returns a new dictionary where each student’s
# name maps to their average grade.
# Sample output: {'Alice': 87.67, 'Bob': 80.0, 'Charlie': 93.33}
def average_grades(grades):
    result = {}
    for name in grades:
        scores = grades[name]

        total = 0
        i = 0
        while i < len(scores):
            total = total + scores[i]
            i += 1

        avg = total / len(scores)
        result[name] = round(avg, 2)

    return result


average_grades(grades = {
"Alice": [85, 90, 88],
"Bob": [78, 82, 80],
"Charlie": [92, 95, 93]
})
print(average_grades)
               


def reverse_lookup(d, value):
    keys = []
    for k in d:
        if d[k] == value:
            keys = keys + [k]
    return keys

d = {'a': 10, 'b': 20, 'c': 10, 'd': 30}
print(reverse_lookup(d, 10))


def merge_dicts(d1, d2):
    merged = {}

    for k in d1:
        merged[k] = d1[k]

    for k in d2:
        if k in merged:
            merged[k] = merged[k] + d2[k]
        else:
            merged[k] = d2[k]

    return merged
d1 = {'a': 10, 'b': 5, 'c': 8}
d2 = {'b': 7, 'c': 3, 'd': 12}
print(merge_dicts(d1, d2))
```
